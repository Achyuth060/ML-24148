```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("CH.SC.U4CSE24148_EX-9.6_2(KNN).csv")

data.head()
```

```python
new_brightness = float(input("Enter new Brightness: "))
new_saturation = float(input("Enter new Saturation: "))
```

```python
data['Distance'] = np.sqrt(
    (data['Brightness'] - new_brightness) ** 2 +
    (data['Saturation'] - new_saturation) ** 2
)

print(data)
```

```python
data = data.sort_values(
    by='Distance'
)

print(data)
```

```python
print("Nearest entry:")
print(data.iloc[0])
```
